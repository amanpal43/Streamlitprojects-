```
import ipywidgets as widgets
from IPython.display import display, clear_output

# Title and Description
title = widgets.HTML("<h2>🎓 Student Information Form</h2>")
description = widgets.HTML("<p>Enter student details and click Submit to view the information.</p>")

# Input Fields
name = widgets.Text(
    description='Name:',
    placeholder='Enter student name'
)

roll = widgets.IntText(
    description='Roll No:'
)

course = widgets.Dropdown(
    options=['BCA', 'BBA', 'MCA', 'MBA', 'B.Tech', 'B.Sc'],
    description='Course:'
)

subjects = widgets.SelectMultiple(
    options=['Python', 'Java', 'C++', 'DBMS', 'AI', 'Machine Learning', 'Data Science'],
    description='Subjects:'
)

marks = widgets.IntSlider(
    value=50,
    min=0,
    max=100,
    step=1,
    description='Marks:'
)

submit_btn = widgets.Button(
    description='Submit',
    button_style='success'
)

output = widgets.Output()

# Submit Function
def submit_form(b):
    with output:
        clear_output()

        # Validation
        if not name.value.strip():
            print("❌ Please enter the student name.")
            return

        if len(subjects.value) == 0:
            print("❌ Please select at least one subject.")
            return

        print("✅ Student details submitted successfully!\n")
        print("📋 Student Details")
        print("-" * 30)
        print(f"Student Name : {name.value}")
        print(f"Roll Number  : {roll.value}")
        print(f"Course       : {course.value}")
        print(f"Subjects     : {', '.join(subjects.value)}")
        print(f"Marks        : {marks.value}")

submit_btn.on_click(submit_form)

# Display Form
display(
    title,
    description,
    name,
    roll,
    course,
    subjects,
    marks,
    submit_btn,
    output
)
```
